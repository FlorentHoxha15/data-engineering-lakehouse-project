# 🥇 Gold Currency — Analytics

## Goal
Transform clean Silver currency data into an analytics-ready Gold table with summary statistics per currency.

## Flow
Silver Delta Table → Aggregation → Gold Delta Table

## 1. Read Silver Data

```python
def read_silver():
    df = spark.table("silver.currency_rates")

    return df
```

## 2. Transform Gold Data

```python
from pyspark.sql.functions import avg, min, max

def transform_gold(df):
    gold_df = df.groupBy(
        "quote"
    ).agg(
        avg("rate").alias("average_rate"),
        min("rate").alias("min_rate"),
        max("rate").alias("max_rate")
    )

    return gold_df
```

## 3. Write to Gold Delta Table

```python
def write_gold(gold_df):
    gold_df.write \
        .format("delta") \
        .mode("overwrite") \
        .saveAsTable("gold.currency_summary")
```

## 4. Run Gold Pipeline

```python
df = read_silver()

gold_df = transform_gold(df)

write_gold(gold_df)
```

```python
%sql
SELECT *
FROM gold.currency_summary;
```
