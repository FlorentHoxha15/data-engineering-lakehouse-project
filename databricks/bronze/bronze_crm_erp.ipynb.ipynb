# 🥉 Bronze CRM & ERP — CSV Ingestion

## Goal
Load raw CRM and ERP CSV source files into Bronze Delta Tables.

## Flow
CRM & ERP CSV Files → PySpark DataFrames → Bronze Delta Tables

## 1. CRM Customer Information

Load raw CRM customer data from the CSV source into the Bronze Delta Table.

```python
crm_cust_df = spark.read \
    .option("header", True) \
    .option("inferSchema", True) \
    .csv("/Workspace/Users/florenthoxha15@gmail.com/sql-data-warehouse-project/datasets/source_crm/cust_info.csv")

display(crm_cust_df)
```

### Load to Bronze Delta Table

```python
crm_cust_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("bronze.crm_cust_info")
```

```python
%sql
SELECT *
FROM bronze.crm_cust_info;
```
